```python
import pandas as pd
```

```python
# Load dataset
df = pd.read_csv('datafinal.csv')
```

```python
# Hapus kolom yang tidak dipakai (hanya pakai sampai kolom label)
kolom_yang_dipakai = ['IDJwb', 'IDPSJ', 'questions', 'answerKeys', 'answer', 'grade']
df = df[kolom_yang_dipakai]
```

```python
# Ringkasan jumlah data per IDPSJ
print("\n" + "=" * 70)
print("Jumlah Sampel per IDPSJ")
print("=" * 70)
print(df['IDPSJ'].value_counts().sort_index())
```

```python
# Proses kolom grade: konversi dari format "10,00" → integer 1-10
# Ganti koma desimal (format Indonesia) dengan titik, lalu konversi ke float → int
df['grade'] = df['grade'].astype(str).str.replace(',', '.').astype(float).round().astype(int)

# Grade 0 diubah menjadi 1 (nilai minimum valid)
df['grade'] = df['grade'].clip(lower=1)

print("Distribusi grade setelah konversi:")
grade_dist = df['grade'].value_counts().sort_index()
print(grade_dist)
```

```python
# Sebaran per IDPSJ (baris=IDPSJ, kolom=grade)
print("\n" + "=" * 70)
print("Sebaran Grade per IDPSJ")
print("=" * 70)
pivot = (df.groupby(['IDPSJ', 'grade'])
           .size()
           .unstack(fill_value=0)
           .reindex(columns=sorted(df['grade'].unique()), fill_value=0))
print(pivot.to_string())
```

```python
# Fitur search berdasarkan IDPSJ dan nilai (grade)
def search_by_idpsj_grade(dataframe, idpsj=None, grade=None):
    """
    Filter dataframe berdasarkan IDPSJ dan/atau grade.
    - idpsj: int atau None
    - grade: int atau None
    """
    result = dataframe.copy()

    if idpsj is not None:
        result = result[result['IDPSJ'] == int(idpsj)]

    if grade is not None:
        result = result[result['grade'] == int(grade)]

    return result


# Contoh pemakaian
idpsj = 12
grade = 10

hasil = search_by_idpsj_grade(df, idpsj=idpsj, grade=grade)

# print(f"Pencarian -> IDPSJ={idpsj}, grade={grade}")
# print(f"Jumlah data ditemukan: {len(hasil)}")

if len(hasil) > 0:
    # Kalau ingin semua kolom, ganti jadi:
    # hasil_list = hasil[['IDJwb', 'IDPSJ', 'questions', 'answerKeys', 'answer', 'grade']].to_dict('records')
    hasil_list = hasil['answer'].tolist()
    print(hasil_list)
else:
    print("Tidak ada data yang cocok.")
```
